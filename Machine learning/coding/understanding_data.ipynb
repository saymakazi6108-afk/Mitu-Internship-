# EDA using univariate analysis

univariate analysis means : 

- “Uni” = one
- You study one column of your dataset
- Goal → understand distribution, pattern, spread

there is two type of data 

1. numerical : height , weight , age etc.
2. categorical : nationality , state etc.

### Categorical data

countplot 


```python
sns.countplot(df['Internship_Experience'])
```

```python
df['Internship_Experience'].value_counts()
```

```python
df['Internship_Experience'].value_counts().plot(kind='bar')
```

```python
sns.countplot(df['Placement'])
```

```python
df['Placement'].value_counts()
```

```python
df['Placement'].value_counts().plot(kind='bar')
```

piechart

```python
df['Internship_Experience'].value_counts().plot(kind='pie')
```

If you want percentage in your piechart .

```python
df['Internship_Experience'].value_counts().plot(kind='pie',autopct='%.2f')
```

#### boxplot
A boxplot is used to quickly understand data using 5 important values:

1. Minimum
2. 25% value (Q1)
3. Median (middle value)
4. 75% value (Q3)
5. Maximum

- used for detect outliers 

```python
sns.boxplot(df['Extra_Curricular_Score'])
```
